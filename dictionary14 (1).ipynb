```python
# Dictionary is Used To Store Multivalue in Variable
#(1).Dictionary Operator
stu1 = {"Name":"Brevis","Age":23,"Game":"Circket"}
stu2 = {"Name":"Jai","Age":23,"Game":"Javelin"}
stu3 = {"Name":"Brevis","Age":20,"Game":"Circket"}
print("Dictionary:",stu1)
#Membership Operator
print("Membership:","Name" in stu1)
#Not Membership Operator
print("Not Membership:","Category" not in stu1)
#Identity Operator
print("Identity:",stu1 is stu3)
#Not Identity Operator
print("Not Identity:",stu1 is not stu2)
#Equality Operator
print("Equality:",stu1 == stu3)
#Inequality Operator
print("Inequality:",stu2!= stu3)
#(2).Dictionary Operation
books ={"Book1":"Power Electronics","Book2":"Gen AI","Book3":"Data Science"}
#Access an Element
print("Access:",books.get("Book1"))
#Update an Element
books["Book2"] = "Machine Learning"
print("Update:",books)
#Append an Element Key With Value
books["Book4"] = "Artificial Intelligence"
print("Append:",books)
#Add Multiple Element Using Update()
books.update({"Book5":"Python","Book6":"Java"})
print("Update:",books)
#Remove an Element Using pop Key with Pair
books.pop("Book6")
print("Pop:",books)
#Iterate Through Dictionay
for key,value in books.items():
    print(key,":",value)
#Remove an Element Using popitem()
books.popitem()
print("Popitem:",books)

#(3).Built-in Function
books ={"Book1":"Power Electronics","Book2":"Gen AI","Book3":"Data Science"}
#Length of Dictionary
print("Length:",len(books))
#Maximum of Dictionary
print("Maximum:",max(books))
#Minimum of Dictionary
print("Minimum:",min(books))
#Diaplay All Keys
print("Keys:",books.keys())
#Display All Values
print("Values:",books.values())
#Display All Items
print("Items:",books.items())
```
